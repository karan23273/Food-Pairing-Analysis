### **Merging**

```python
# -----------------------------
# Normalize text
# -----------------------------
def norm(x):
    return x.lower().strip() if isinstance(x, str) else x

recipes = df1.rename(columns={
    "Recipe_id": "recipe_id",
    "Region": "cuisine",
    "Sub_region": "country",
    "Continent": "continent"
})

ingredients = df2.rename(columns={
    "IngID": "entity_id",
    "generic_name": "aliased_ingredient_name",
    "Dietrx_Category": "category"
})

aliases = df3.rename(columns={
    "Recipe_Number": "recipe_id",
    "ing_id": "entity_id"
})

recipes["cuisine"] = recipes["cuisine"].apply(norm)
recipes["country"] = recipes["country"].apply(norm)
recipes["continent"] = recipes["continent"].apply(norm)

ingredients["aliased_ingredient_name"] = ingredients["aliased_ingredient_name"].apply(norm)
ingredients["category"] = ingredients["category"].apply(norm)


# -----------------------------
# Ingredient → Category mapping via Entity ID Joins
# -----------------------------
aliases = aliases.merge(
    ingredients[["entity_id", "aliased_ingredient_name", "category"]],
    on="entity_id",
    how="left"
 )

# Drop ingredients without category (noise)
aliases = aliases.dropna(subset=["category"])

# Attach cuisine
# -----------------------------
edgelist = aliases.merge(
    recipes[["recipe_id", "cuisine"]],
    on="recipe_id",
    how="left"
 )

# -----------------------------
# Final edgelist
# -----------------------------
edgelist = edgelist[
    ["recipe_id", "aliased_ingredient_name", "cuisine", "category"]
].sort_values("recipe_id")

# Ensure one edge per recipe + ingredient
edgelist = edgelist.drop_duplicates(subset=["recipe_id", "aliased_ingredient_name"])

# -----------------------------
# Save
# -----------------------------
edgelist.to_csv("ingredient_edgelist.csv", index=False)
# edgelist.sort_values("recipe_id").head().
# RECIPE ID STARTS from 2610
edgelist.head(20)
```

```python
# Number of recipes per cuisine
recipes_per_cuisine = (
    df[["cuisine", "recipe_id"]]
    .drop_duplicates()
    .groupby("cuisine")
    .size()
    .rename("num_recipes")
)

# Number of unique ingredients per cuisine
ingredients_per_cuisine = (
    df[["cuisine", "aliased_ingredient_name"]]
    .drop_duplicates()
    .groupby("cuisine")
    .size()
    .rename("num_ingredients")
)

# Combine
stats = pd.concat(
    [recipes_per_cuisine, ingredients_per_cuisine],
    axis=1
).reset_index()

stats
```

```python
x = stats["num_ingredients"].values
y = stats["num_recipes"].values

# Linear fit in log-space
coeffs = np.polyfit(x, np.log10(y), 1)
a, b = coeffs

# Generate smooth curve
x_fit = np.linspace(x.min(), x.max(), 200)
y_fit = 10 ** (a * x_fit + b)
```

```python
plt.figure(figsize=(12, 8))

# Scatter points
plt.scatter(
    stats["num_ingredients"],
    stats["num_recipes"],
    s=80,
    alpha=0.9
)

# Exponential trend
plt.plot(
    x_fit,
    y_fit,
    linestyle="--",
    color="gray",
    linewidth=2
)

# Log scale on y-axis
plt.yscale("log")

# Labels
plt.xlabel("Total Number of Ingredients", fontsize=16)
plt.ylabel("Number of Recipes", fontsize=16)

# Annotate cuisines
for _, row in stats.iterrows():
    plt.text(
        row["num_ingredients"] + 5,
        row["num_recipes"],
        row["cuisine"].title(),
        fontsize=10
    )

plt.tight_layout()
plt.show()
```

```python
import pickle
import os
import csv
import networkx as nx
from collections import defaultdict


# ---------------------------------------------------------------------
# Configuration
# ---------------------------------------------------------------------

# Location of the raw input data directory.
data_loc = ""

# Initialize a dictionary keyed by cuisine -> NetworkX MultiGraph.
# Using defaultdict with nx.MultiGraph as the factory ensures a new empty
# MultiGraph is created upon first access for a missing cuisine key.
nx_graph_dict = defaultdict(nx.MultiGraph)

# ---------------------------------------------------------------------
# Ingest raw edgelist and build bipartite graphs per cuisine
# ---------------------------------------------------------------------

# Read recipe–ingredient edges from the semicolon-separated CSV.
with open(os.path.join(data_loc, "ingredient_edgelist.csv"), "r") as f:
    reader = csv.reader(f)
    next(reader)  # skip header

    for line in reader:
        # The file is semicolon-separated; we split explicitly.
        # The file is actually comma-separated, so we can use `line` directly.
        recipe_id = int(line[0])
        ingredient = line[1]
        cuisine = line[2]
        ing_type = line[3]

        # If this is the first time we see the cuisine, create its graph and
        # add the current nodes/edge. Otherwise, ensure nodes exist and add edge.
        if cuisine not in nx_graph_dict.keys():
            nx_graph_dict[cuisine] = nx.MultiGraph()

            # Add recipe node with attributes: bipartite partition and cuisine.
            nx_graph_dict[cuisine].add_node(
                recipe_id, bipartite=0, cuisine=cuisine
            )

            # Add ingredient node with attributes: bipartite partition, display name, and category.
            nx_graph_dict[cuisine].add_node(
                ingredient, bipartite=1, name=ingredient, category=ing_type
            )

            # Connect recipe and ingredient with an edge.
            nx_graph_dict[cuisine].add_edge(recipe_id, ingredient)

        else:
            # Ensure the recipe node exists (idempotent).
            if not nx_graph_dict[cuisine].has_node(recipe_id):
                nx_graph_dict[cuisine].add_node(
                    recipe_id, bipartite=0, cuisine=cuisine
                )

            # Ensure the ingredient node exists (idempotent).
            if not nx_graph_dict[cuisine].has_node(ingredient):
                nx_graph_dict[cuisine].add_node(
                    ingredient, bipartite=1, name=ingredient, category=ing_type
                )

            # Connect recipe and ingredient with an edge.
            nx_graph_dict[cuisine].add_edge(recipe_id, ingredient)

# ---------------------------------------------------------------------
# Persist outputs: one GraphML per cuisine, plus a pickle with all graphs
# ---------------------------------------------------------------------

# Ensure output directories exist
graphml_dir = "post_processed_data/graphml_food_nets_directory"
os.makedirs(graphml_dir, exist_ok=True)

pickle_dir = "post_processed_data"
os.makedirs(pickle_dir, exist_ok=True)

for cuisine in nx_graph_dict.keys():
    graph = nx_graph_dict[cuisine]
    # Writes a GraphML file named after the cuisine.
    nx.write_graphml_lxml(
        graph, f"{graphml_dir}/{cuisine}.graphml"
    )

# Save the full dictionary of graphs for convenient re-use.
pickle.dump(nx_graph_dict, open(f"{pickle_dir}/graph_dict_from_edgelist.pkl", "wb"))
```

```python
"""
Ridge plot of recipe-size distributions across cuisines.

This script loads the per-cuisine recipe–ingredient bipartite graphs and computes,
for each cuisine:
- Recipe size (number of ingredients per recipe)
- Recipe *type* size (number of distinct ingredient categories per recipe)

It then produces a ridge (joy) plot showing the kernel density estimate (KDE)
of recipe sizes for each cuisine, colored by continent.

Inputs
------
- post_processed_data/graph_dict_from_edgelist.pkl
    A dict: cuisine -> NetworkX MultiGraph (bipartite). Nodes:
      - recipes: bipartite=0
      - ingredients: bipartite=1 with node attribute 'category'

Outputs
-------
- figures/basic_statistics/recipe_size_dist_ridgeplot.pdf

Notes
-----
- Excludes cuisines: misc_central_american, misc_dutch, misc_belgian
- Cuisines are ordered by their mean recipe size (ascending).
- Seaborn FacetGrid is used to build the ridgeline chart (one KDE per row).
"""

import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx  # kept for environment parity (graphs are NetworkX)


# ---------------------------------------------------------------------
# Load bipartite graphs (repo-friendly path) and filter cuisines
# ---------------------------------------------------------------------
graph_dict = pickle.load(open("post_processed_data/graph_dict_from_edgelist.pkl", "rb"))

distributions = []
type_distributions = []

exclude_list = ["misc_central_american", "misc_dutch", "misc_belgian"]
updated_dict = {key: value for key, value in graph_dict.items() if key not in exclude_list}

# ---------------------------------------------------------------------
# Compute per-cuisine recipe size and type size lists
# ---------------------------------------------------------------------
for cuisine in updated_dict.keys():
    cuisine_graph = updated_dict[cuisine]

    recipes = {n for n, d in cuisine_graph.nodes(data=True) if d["bipartite"] == 0}
    ingredient_type_dict = {n: d["category"] for n, d in cuisine_graph.nodes(data=True) if d["bipartite"] == 1}

    # Recipe sizes (ingredient counts)
    degree_sequence = sorted((d for n, d in cuisine_graph.degree() if n in recipes), reverse=True)

    # Recipe type sizes (distinct category counts)
    type_degree_sequence = []
    for recipe in recipes:
        type_degree_sequence.append(len(set(ingredient_type_dict[ing] for ing in cuisine_graph.neighbors(recipe))))

    distributions.append(degree_sequence)
    type_distributions.append(type_degree_sequence)

labels = list(updated_dict.keys())

# ---------------------------------------------------------------------
# Build long-form DataFrame with summary columns for ordering
# ---------------------------------------------------------------------
data = []
for i, (dist, type_dist) in enumerate(zip(distributions, type_distributions)):
    for value1, value2 in zip(dist, type_dist):
        data.append(
            [
                value1,                 # recipe_size
                value2,                 # recipe_type_size
                labels[i],              # cuisine
                max(dist),              # max_recipe_size
                np.mean(dist),          # mean_recipe_size
                max(type_dist),         # max_type_size
                np.mean(type_dist),     # mean_type_size
            ]
        )

df = pd.DataFrame(
    data,
    columns=[
        "recipe_size",
        "recipe_type_size",
        "cuisine",
        "max_recipe_size",
        "mean_recipe_size",
        "max_type_size",
        "mean_type_size",
    ],
).sort_values(by=["mean_recipe_size"])

# Optional hatch groups (left in-place; final plot uses black edges instead)
hatching_groups = [
    list(updated_dict.keys())[:6],
    list(updated_dict.keys())[6:12],
    list(updated_dict.keys())[12:18],
    list(updated_dict.keys())[18:],
]
hatches = ["|", "-", "/", "\\ "]
df["hatch"] = [""] * len(df)
for group, hatch in zip(hatching_groups, hatches):
    index = df.index[df["cuisine"].isin(group)]
    for idx in index:
        df.loc[idx, "hatch"] = hatch

# ---------------------------------------------------------------------
# Colors by continent (aligned to graph cuisine keys)
# ---------------------------------------------------------------------
cuisine_to_continents = {
    "australian": "Oceania",
    "belgian": "Europe",
    "canadian": "America",
    "caribbean": "America",
    "central american": "America",
    "chinese and mongolian": "Asia",
    "deutschland": "Europe",
    "eastern european": "Europe",
    "french": "Europe",
    "greek": "Europe",
    "indian subcontinent": "Asia",
    "irish": "Europe",
    "italian": "Europe",
    "japanese": "Asia",
    "korean": "Asia",
    "mexican": "America",
    "middle eastern": "Asia",
    "northern africa": "Africa",
    "rest africa": "Africa",
    "scandinavian": "Europe",
    "south american": "America",
    "southeast asian": "Asia",
    "spanish and portuguese": "Europe",
    "thai": "Asia",
    "uk": "Europe",
    "us": "America",
}
continents_to_colors = {"Africa": "#b2df8a", "America": "#fb9a99", "Asia": "#fdbf6f", "Europe": "#a6cee3", "Oceania": "#cab2d6"}

missing_cuisines = [cuisine for cuisine in df["cuisine"].unique() if cuisine not in cuisine_to_continents]
if missing_cuisines:
    raise ValueError(f"Missing continent mapping for cuisines: {missing_cuisines}")

colordict = {
    cuisine: continents_to_colors[cuisine_to_continents[cuisine]]
    for cuisine in df["cuisine"].unique()
}

# ---------------------------------------------------------------------
# Plot: ridgeline KDE of recipe sizes by cuisine
# ---------------------------------------------------------------------
sns.set_theme(style="ticks", rc={"axes.facecolor": (0, 0, 0, 0), "axes.linewidth": 2, "font.family": "DejaVu Sans"})
sns.set_context(context="paper")

g = sns.FacetGrid(df, row="cuisine", hue="cuisine", aspect=9, height=1.2, palette=colordict)

# Filled KDE per row
g.map_dataframe(sns.kdeplot, x="recipe_size", fill=True, alpha=1, cut=0)

# Per-row label at right
def label(color, label):
    label = label.replace("_", " ").title()
    ax = plt.gca()
    ax.text(1, 0.2, label, color=color, fontsize=20, ha="right", va="top", transform=ax.transAxes)

g.map(label, label="cuisine", color="black")

# Layout and axes cosmetics
g.fig.subplots_adjust(hspace=0)
g.set_titles("")
g.set(yticks=[], xticks=np.arange(0, 30, 5), xlabel="recipe_size")
g.set_xticklabels(np.arange(0, 30, 5), size=28)
g.set(xlim=(0, 25))
g.despine(left=True)
g.set(ylabel="")

# Add black edges to filled KDE patches for clarity
for ax in g.axes.flat:
    for patch in ax.collections:
        patch.set_edgecolor("black")

plt.xlabel("Recipe Size", fontsize=28, labelpad=25)

# ---------------------------------------------------------------------
# Save figure (repo-friendly path)
# ---------------------------------------------------------------------
plt.savefig("recipe_size_dist_ridgeplot.pdf", bbox_inches="tight")
plt.show()
```

```python
"""
Extract ingredient-type (food category) graphs from recipe–ingredient bipartite graphs.

This script builds *ingredient-category co-occurrence graphs* for each cuisine,
starting from the recipe–ingredient bipartite graphs previously generated and saved
with `build_bipartite_from_edgelist.py`.

For each cuisine:
- Nodes = ingredient categories (e.g., "dairy", "spice", "grain").
- Weighted edges = number of recipes where two categories co-occur.
- Node attribute `counts` = fraction of recipes in which that category appears.


Input
-----
- Pickled bipartite graph dictionary:
  `post_processed_data/graph_dict_from_edgelist.pkl`

  Each entry is a `networkx.MultiGraph` with:
  - bipartite=0 nodes: recipes
  - bipartite=1 nodes: ingredients (with `category` attribute)

Output
------
- Per-cuisine GraphML files:
  `post_processed_data/graphml_food_type_nets_directory/<cuisine>.graphml`
- A pickle with all cuisine category graphs:
  `post_processed_data/nx_category_graph_dict_from_edgelist.pickle`

Graph Schema (per cuisine)
--------------------------
- Graph type: `networkx.Graph` (simple weighted graph)
- Nodes: ingredient categories
    - attributes:
        - `counts` : float, fraction of recipes in the cuisine containing at least
                     one ingredient of that category.
- Edges: weighted by the number of co-occurrences across recipes.

Notes
-----
- Intermediate `MultiGraph` objects are used to count category co-occurrences
  per recipe before aggregation.
- Edge weights in the final graph represent the *total number of recipes*
  where two categories co-occur, summed over all recipes in the cuisine.
- The script assumes that `post_processed_data/graphml_food_type_nets_directory/`
  exists; otherwise, `nx.write_graphml_lxml` will fail.

Example
-------
Run after building the bipartite graphs:

    python scripts/build_category_graphs.py

"""

import pickle
import numpy as np
import networkx as nx
import os


# ---------------------------------------------------------------------
# Load bipartite recipe–ingredient graphs
# ---------------------------------------------------------------------

graph_dict = pickle.load(
    open("post_processed_data/graph_dict_from_edgelist.pkl", "rb")
)

# Extract the unique set of ingredient categories across all cuisines
ing_categories = np.unique(
    np.hstack(
        [
            [
                d["category"]
                for n, d in graph_dict[cuisine].nodes(data=True)
                if d["bipartite"] == 1
            ]
            for cuisine in graph_dict.keys()
        ]
    )
)

# Dictionary: cuisine -> category co-occurrence graph
nx_category_graph_dict = {}

# ---------------------------------------------------------------------
# Build category co-occurrence graphs
# ---------------------------------------------------------------------

for cuisine in graph_dict.keys():
    try:
        cuisine_graph = graph_dict[cuisine]

        # Initialize category graph for this cuisine
        category_graph = nx.MultiGraph()
        category_graph.add_nodes_from(ing_categories)

        # Counter for per-category recipe presence
        counter_dict = {key: 0 for key in ing_categories}

        # Separate recipes and ingredient-category mapping
        recipes = {n for n, d in cuisine_graph.nodes(data=True) if d["bipartite"] == 0}
        ingredients = {
            n: d["category"]
            for n, d in cuisine_graph.nodes(data=True)
            if d["bipartite"] == 1
        }

        # Process each recipe: connect its ingredient categories
        for recipe in recipes:
            category_list = [
                ingredients[key] for key in list(cuisine_graph.neighbors(recipe))
            ]

            # Build a temporary multigraph for this recipe’s categories
            g = nx.MultiGraph()
            g.add_nodes_from(category_list)

            for idx in range(len(category_list)):
                node = category_list[idx]
                for n in category_list[idx + 1 :]:
                    g.add_edge(node, n)

            # Convert to weighted simple graph
            g_weighted = nx.Graph()
            for u, v, d in g.edges(data=True):
                weight = g.number_of_edges(u, v)
                g_weighted.add_edge(u, v, weight=weight)

            # Add edges (weight=1 per co-occurrence) to the cuisine category graph
            category_graph.add_weighted_edges_from(
                (node1, node2, w)
                for (node1, node2), w in zip(
                    g_weighted.edges(), [1 for _ in range(len(g_weighted.edges()))]
                )
            )

            # Update category presence counts (normalized by number of recipes)
            for cat in set(category_list):
                counter_dict[cat] += 1 / len(recipes)

        # Aggregate weighted multigraph into a simple weighted graph
        G = nx.Graph()
        G.add_nodes_from(ing_categories)
        for u, v, data in category_graph.edges(data=True):
            w = data["weight"]
            if G.has_edge(u, v):
                G[u][v]["weight"] += w
            else:
                G.add_edge(u, v, weight=w)

        # Attach per-category recipe frequency counts
        nx.set_node_attributes(G, counter_dict, "counts")

        # Store final graph
        nx_category_graph_dict[cuisine] = G

    except KeyError:
        # Skip cuisines missing expected attributes
        pass

# ---------------------------------------------------------------------
# Persist outputs: GraphML per cuisine + pickle of all graphs
# ---------------------------------------------------------------------

graphml_dir = "post_processed_data/graphml_food_type_nets_directory"
os.makedirs(graphml_dir, exist_ok=True)

for cuisine in nx_category_graph_dict.keys():
    ig_category_graph = nx_category_graph_dict[cuisine]
    nx.write_graphml_lxml(
        ig_category_graph,
        f"{graphml_dir}/{cuisine}.graphml",
    )

os.makedirs("post_processed_data", exist_ok=True)

pickle.dump(
    nx_category_graph_dict,
    open("post_processed_data/nx_category_graph_dict_from_edgelist.pickle", "wb"),
)
```

```python
import pickle
import pandas as pd
import itertools
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import os

# ---------------------------------------------------------------------
# Caption (as requested)
# ---------------------------------------------------------------------
print(
    "Strip plots of z-scores for edge weights across world cuisines.\n"
    "For each cuisine on the y-axis, the x-axis shows the z-scores of edge weights relative to the distribution of the corresponding edge weights across all 26 cuisines. A jitter is applied for better visualization.\n\n"
    "Two edges are highlighted for each cuisine: one with a high z-score and one with a low z-score, representing significant deviations in the co-occurrence patterns."
 )

# ---------------------------------------------------------------------
# Cuisine order/labels + continent mapping (from Cell 71)
# ---------------------------------------------------------------------
main_df = pd.read_csv("ingredient_edgelist.csv")
main_df["cuisine"] = main_df["cuisine"].astype(str).str.strip().str.lower()

cuisines = sorted(set(main_df["cuisine"].unique()))
label_overrides = {
    "chinese and mongolian": "Chinese & Mongolian",
    "deutschland": "German",
    "eastern european": "East European",
    "indian subcontinent": "Indian",
    "middle eastern": "Middle Eastern",
    "northern africa": "Northern Africa",
    "rest africa": "Rest Africa",
    "south american": "South American",
    "southeast asian": "Southeast Asian",
    "spanish and portuguese": "Spanish & Portuguese",
    "uk": "UK",
    "us": "American",
}
cuisine_label_pairs = sorted(
    [(c, label_overrides.get(c, c.title())) for c in cuisines],
    key=lambda x: x[1].lower(),
 )
cuisines = [c for c, _ in cuisine_label_pairs]
labels = [label for _, label in cuisine_label_pairs]
label_to_y = {label: i for i, label in enumerate(labels)}

cuisine_to_continents = {
    "australian": "Oceania",
    "belgian": "Europe",
    "canadian": "America",
    "caribbean": "America",
    "central american": "America",
    "chinese and mongolian": "Asia",
    "deutschland": "Europe",
    "eastern european": "Europe",
    "french": "Europe",
    "greek": "Europe",
    "indian subcontinent": "Asia",
    "irish": "Europe",
    "italian": "Europe",
    "japanese": "Asia",
    "korean": "Asia",
    "mexican": "America",
    "middle eastern": "Asia",
    "northern africa": "Africa",
    "rest africa": "Africa",
    "scandinavian": "Europe",
    "south american": "America",
    "southeast asian": "Asia",
    "spanish and portuguese": "Europe",
    "thai": "Asia",
    "uk": "Europe",
    "us": "America",
}
continents_to_colors = {
    "Africa": "#b2df8a",
    "America": "#fb9a99",
    "Asia": "#fdbf6f",
    "Europe": "#a6cee3",
    "Oceania": "#cab2d6",
}
missing_in_map = [c for c in cuisines if c not in cuisine_to_continents]
if missing_in_map:
    raise ValueError(f"Missing continent mapping for cuisines: {missing_in_map}")
cuisine_to_colors = {
    cuisine: continents_to_colors[cuisine_to_continents[cuisine]]
    for cuisine in cuisines
}

# ---------------------------------------------------------------------
# Load graphs (weights per cuisine)
# ---------------------------------------------------------------------
graphs = pickle.load(open("post_processed_data/nx_category_graph_dict_from_edgelist.pickle", "rb"))
graphs = {str(k).strip().lower(): v for k, v in graphs.items()}

# -------------------------------
# Pair counts (RAW graphs)
# -------------------------------
raw_graphs = pickle.load(open("post_processed_data/nx_category_graph_dict_from_edgelist.pickle", "rb"))
raw_graphs = {str(k).strip().lower(): v for k, v in raw_graphs.items()}

pair_counts = {}

for cuisine, G in raw_graphs.items():
    pair_counts[cuisine] = {}
    for u, v, d in G.edges(data=True):
        pair = tuple(sorted((u, v)))
        pair_counts[cuisine][pair] = d["weight"]

missing_graphs = [c for c in cuisines if c not in graphs]
if missing_graphs:
    raise ValueError(f"Missing graphs for cuisines: {missing_graphs}")

# ---------------------------------------------------------------------
# Build z-score dict for every edge across cuisines (26 cuisines)
# ---------------------------------------------------------------------
ingredient_types = [
    "Additive","Bakery","Beverage","Beverage Alcoholic","Cereal","Condiment","Dairy","Dish",
    "Essential Oil","Fish","Flower","Fruit","Fungus","Herb","Legume","Maize","Meat","Nuts And Seeds",
    "Plant","Seafood","Spice","Vegetable",
]

edges = set()

for G in graphs.values():
    for u, v in G.edges():
        edges.add(tuple(sorted((u, v))))

edges = list(edges)
zscore_dict = {}
for edge in edges:
    weights = []
    for cuisine in cuisines:
        G = graphs[cuisine]
        weights.append(G[edge[0]][edge[1]]["weight"] if G.has_edge(*edge) else 0)
    mean_weight = np.mean(weights)
    std_weight = np.std(weights)
    z_scores = [0 for _ in weights] if std_weight == 0 else [(w - mean_weight) / std_weight for w in weights]
    zscore_dict[edge] = z_scores

# ---------------------------------------------------------------------
# Long-form DataFrame: one row per cuisine x edge
# ---------------------------------------------------------------------
data_list = []
for i, cuisine in enumerate(cuisines):
    for edge, z_scores in zscore_dict.items():
        data_list.append([cuisine, labels[i], cuisine_to_continents[cuisine], edge, z_scores[i]])

edge_df = pd.DataFrame(data_list, columns=["Cuisine", "Cuisine Label", "Continent", "Edge", "Z-Score"])

# ---------------------------------------------------------------------
# Choose highlights automatically: per cuisine, max and min z-score edge
# ---------------------------------------------------------------------
edge_df["Highlight"] = False
edge_df["Info"] = ""
highlight_rows = []
for cuisine in cuisines:
    sub = edge_df[edge_df["Cuisine"] == cuisine]
    if sub.empty:
        continue
    max_idx = sub["Z-Score"].idxmax()
    min_idx = sub["Z-Score"].idxmin()
    for idx, kind in [(max_idx, "high"), (min_idx, "low")]:
        edge_df.loc[idx, "Highlight"] = True
        e = edge_df.loc[idx, "Edge"]
        pair = tuple(sorted(e))
        count = int(pair_counts[cuisine].get(pair, 0))
        info = f"{e[0]}-{e[1]} ({count})"
        edge_df.loc[idx, "Info"] = info
        highlight_rows.append((cuisine, kind, info, float(edge_df.loc[idx, "Z-Score"])))

highlighted = edge_df[edge_df["Highlight"]].copy()

# ---------------------------------------------------------------------
# Plot
# ---------------------------------------------------------------------
def draw_background_regions(ax, Delta_tau, t_min, t_max, color, alpha):
    t_old = t_min
    t_next = t_old + Delta_tau
    while t_old < t_max:
        ax.axvspan(t_old, t_next, ymin=0, ymax=1, color=color, alpha=alpha, zorder=1)
        t_old = t_old + 2 * Delta_tau
        t_next = t_next + 2 * Delta_tau

fig, ax = plt.subplots(figsize=(21, len(cuisines) * 2.5))

# Base strip plot (all edges)
sns.stripplot(
    x="Z-Score",
    y="Cuisine Label",
    data=edge_df,
    size=6,
    hue="Cuisine",
    palette=cuisine_to_colors,
    jitter=0.175,
    alpha=0.25,
    ax=ax,
    legend=False,
 )

# Overlay highlights
sns.stripplot(
    x="Z-Score",
    y="Cuisine Label",
    data=highlighted,
    size=7,
    hue="Cuisine",
    palette=cuisine_to_colors,
    jitter=0.175,
    alpha=1,
    edgecolor="black",
    linewidth=1.5,
    ax=ax,
    legend=False,
 )

# Annotate: exactly two per cuisine, offset up/down to reduce overlap
for cuisine, kind, info, z in highlight_rows:
    y = label_to_y[labels[cuisines.index(cuisine)]]
    dy = 0.18 if kind == "high" else -0.18
    dx = 0.35 if z >= 0 else -0.35
    ha = "left" if z >= 0 else "right"
    ax.annotate(
        info,
        xy=(z, y),
        xytext=(z + dx, y + dy),
        arrowprops=dict(facecolor="black", width=0.8, headwidth=4, alpha=0.6, connectionstyle="arc3,rad=0"),
        fontsize=12,
        ha=ha,
        va="center",
)

draw_background_regions(ax, Delta_tau=1, t_min=-4, t_max=-1, color="lightgray", alpha=0.25)
draw_background_regions(ax, Delta_tau=1, t_min=1, t_max=5, color="lightgray", alpha=0.25)

ax.axvline(x=0, ymin=0, ymax=1, alpha=1, color="black", linewidth=1, zorder=0)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.set_yticks(range(len(labels)))
ax.set_yticklabels(labels)
ax.tick_params(axis="both", which="major", labelsize=20)
ax.set_xlabel("Z-Score", fontsize=22)
ax.set_ylabel("Cuisine", fontsize=22)

os.makedirs("figures/type_graphs", exist_ok=True)
plt.savefig("figures/type_graphs/z_score_stripplot.svg", bbox_inches="tight")
plt.show()
```

```python
"""
Generate Maximum Spanning Trees (MSTs) from ingredient-type (category) graphs.

This script:
1) Loads per-cuisine category co-occurrence graphs (type graphs).
2) Normalizes edge weights within each graph so they sum to 1.
3) Computes the Maximum Spanning Tree (MST) for each cuisine graph.
4) Saves a dictionary {cuisine: MST graph} to disk.

Inputs
------
- post_processed_data/nx_category_graph_dict_from_edgelist.pickle
    Dict: cuisine -> NetworkX weighted Graph (edge attribute 'weight').

Outputs
-------
- post_processed_data/max_spanning_tree_dict.pickle
    Dict: cuisine -> NetworkX Graph (MST).

Notes
-----
- Behavior preserved except:
  * Removed Minimum Spanning Tree generation as requested.
  * Removed unused imports.
  * Paths updated to repo-friendly locations.
"""

import os
import pickle
import networkx as nx


# ---------------------------------------------------------------------
# Helper: normalize weights in-place so total edge weight = 1 per graph
# ---------------------------------------------------------------------
def normalize_weights(graph: nx.Graph) -> None:
    total_weight = sum(weight for _, _, weight in graph.edges.data("weight"))
    for u, v, data in graph.edges(data=True):
        data["weight"] /= total_weight


# ---------------------------------------------------------------------
# Load category graphs
# ---------------------------------------------------------------------
input_path = "post_processed_data/nx_category_graph_dict_from_edgelist.pickle"
with open(input_path, "rb") as f:
    category_graph_dict = pickle.load(f)

# ---------------------------------------------------------------------
# Build MSTs per cuisine
# ---------------------------------------------------------------------
max_span_tree_dict = {}

for cuisine, graph in category_graph_dict.items():
    # Normalize weights before computing MST
    normalize_weights(graph)

    # Compute Maximum Spanning Tree using edge attribute 'weight'
    max_spanning_tree = nx.maximum_spanning_tree(graph, algorithm="kruskal", weight="weight")
    max_span_tree_dict[cuisine] = max_spanning_tree

# ---------------------------------------------------------------------
# Save results
# ---------------------------------------------------------------------
os.makedirs("post_processed_data", exist_ok=True)
with open("post_processed_data/max_spanning_tree_dict.pickle", "wb") as f:
    pickle.dump(max_span_tree_dict, f)
```

```python
"""
Plot Maximum Spanning Trees (MSTs) of ingredient-type graphs per cuisine.

This script loads:
1) The recipe–ingredient bipartite graphs (for category frequency sizing).
2) A precomputed dictionary of per-cuisine Maximum Spanning Trees (MSTs)
   of ingredient-type co-occurrence graphs.

For each cuisine, it:
- Computes node sizes from category frequencies across recipes.
- Draws the MST with category-colored nodes and weighted edges.
- Places category SVG icons near nodes along incident edges.
- Saves an SVG figure per cuisine.


Inputs
------
- post_processed_data/graph_dict_from_edgelist.pkl
    A dict: cuisine -> NetworkX MultiGraph (bipartite), with ingredients (bipartite=1)
    carrying node attribute 'category'.
- post_processed_data/max_spanning_tree_dict.pickle
    A dict: cuisine -> NetworkX Graph (MST) with edge attribute 'weight'.
- figures/category_symbols/<Category>.svg
    SVG icon for each ingredient category key listed in `color_dict`.

Outputs
-------
- figures/type_graphs/max_spanning_trees/<cuisine>.svg

Notes
-----
- The MST positions are generated using graphviz 'twopi' layout via pygraphviz.
- Node sizes are proportional to per-category frequency across all recipes
  in the corresponding cuisine bipartite graph.
- Ensure the output directory exists before running this script.
"""

import os
import io
import pickle
import pygraphviz as pgv  # required for 'twopi' layout through networkx-agraph
import cairosvg
# import seaborn as sns  # kept for environment parity, not used directly
import networkx as nx
import numpy as np
from matplotlib import pyplot as plt
# from adjustText import adjust_text  # kept for environment parity, not used directly
# from matplotlib.textpath import TextToPath  # kept for environment parity, not used directly
# from matplotlib.text import Text  # kept for environment parity, not used directly
# from matplotlib.patheffects import withStroke  # kept for environment parity, not used directly
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
# from svg.path import parse_path  # kept for environment parity, not used directly
# from matplotlib.bezier import BezierSegment  # kept for environment parity, not used directly

# ---------------------------------------------------------------------
# Visualization palette: category -> color
# ---------------------------------------------------------------------
color_dict = {
    "Fish": "#1F78B4",
    "Meat": "#c44240",
    "Legume": "#E6AB02",
    "Seafood": "#A6CEE3",
    "Fruit": "#f97306",
    "Cereal": "#FDBF6F",
    "Vegetable": "#33A02C",
    "Herb": "#B2Df8A",
    "Fungus": "#BA9E88",
    "Plant": "#1B9E77",
    "Bakery": "#FFFF99",
    "Maize": "#F4D054",
    "Flower": "#fe86a4",
    "Nuts And Seeds": "#645403",
    "Beverage Alcoholic": "#8c0034",
    "Beverage": "#CAB2D6",
    "Additive": "#d3d3d3",
    "Essential Oil": "#677A04",
    "Spice": "#C49C94",
    "Dairy": "#FFF39A",
    "Condiment": "#FF7F00",
    "Dish": "#6A3D9A"
}

# ---------------------------------------------------------------------
# Icon directory (repo-friendly path). One SVG per category name key above.
# ---------------------------------------------------------------------
svg_directory = "category_symbols/"
def normalize(x):
    return x.strip().lower().replace(" ", "_")

# Preload SVG file paths for categories
category_svg_dict = {}
for category in color_dict.keys():
    svg_file_path = os.path.join(svg_directory, f"{category}.svg")
    if os.path.exists(svg_file_path):
        category_svg_dict[normalize(category)] = svg_file_path

# ---------------------------------------------------------------------
# Load data (repo-friendly paths)
# ---------------------------------------------------------------------
# Bipartite cuisine graphs (recipes ↔ ingredients with 'category' on ingredients)
graph_dict = pickle.load(open("post_processed_data/graph_dict_from_edgelist.pkl", "rb"))

# Precomputed MSTs of ingredient-type graphs
max_span_tree_dict = pickle.load(open("post_processed_data/max_spanning_tree_dict.pickle", "rb"))

# Collect the set of all ingredient categories across cuisines
ing_categories = np.unique(
    np.hstack(
        [
            [d["category"] for n, d in graph_dict[cuisine].nodes(data=True) if d["bipartite"] == 1]
            for cuisine in graph_dict.keys()
        ]
    )
)
nx_category_graph_dict = {}

# ---------------------------------------------------------------------
# Plot per-cuisine MSTs
# ---------------------------------------------------------------------
for cuisine in graph_dict.keys():
    fig, ax = plt.subplots(figsize=(15, 15))

    max_span_tree = max_span_tree_dict[cuisine]
    cuisine_graph = graph_dict[cuisine]

    # Count category occurrences across recipes (for node sizing)
    counter_dict = {key: 0 for key in ing_categories}

    recipes = {n for n, d in cuisine_graph.nodes(data=True) if d["bipartite"] == 0}
    ingredients = {n: d["category"] for n, d in cuisine_graph.nodes(data=True) if d["bipartite"] == 1}

    for recipe in recipes:
        category_list = list(ingredients[key] for key in list(cuisine_graph.neighbors(recipe)))
        for cat in category_list:
            counter_dict[cat] += 1

    norm_factor = sum(list(counter_dict.values()))
    freq_dict = {key: val / norm_factor for key, val in counter_dict.items()}

    # Node sizes proportional to category frequency
    node_sizes = np.array(list(freq_dict[node] for node in max_span_tree.nodes())) * 5000
    sel_weights = nx.get_edge_attributes(max_span_tree, "weight").values()

    # Use Graphviz 'twopi' (radial) layout via pygraphviz
    graph_dot = nx.drawing.nx_agraph.to_agraph(max_span_tree)
    graph_dot.layout(prog="twopi")
    pos = {
        n: (
            float(graph_dot.get_node(n).attr["pos"].split(",")[0]),
            float(graph_dot.get_node(n).attr["pos"].split(",")[1]),
        )
        for n in graph_dot.nodes()
    }

    # Node colors ordered by node iteration
    ordered_colors = [
        color_dict.get(category.title(), "#999999")  # fallback gray
        for category in max_span_tree.nodes()
    ]
    # Draw nodes and edges
    nx.draw_networkx_nodes(
        max_span_tree,
        pos,
        ax=ax,
        label=True,
        node_size=node_sizes,
        alpha=1,
        node_color=ordered_colors,
        edgecolors="black",
        linewidths=2,
    )
  #   nx.draw_networkx_labels(
  #   max_span_tree,
  #   pos,
  #   font_size=14,
  #   font_color="black"
  # )

    nx.draw_networkx_edges(
        max_span_tree,
        pos,
        ax=ax,
        width=300 * np.array(list(sel_weights), dtype="float64"),
        edge_color="grey",
        alpha=0.6,
    )

    # Place category icons near nodes along incident edges
    drawn_nodes = set()

    for node in max_span_tree.nodes():

        if node in drawn_nodes:
            continue

        x, y = pos[node]

        svg_file_path = category_svg_dict.get(normalize(node))

        if svg_file_path:
            png_data = cairosvg.svg2png(url=svg_file_path, output_width=50, output_height=50)
            image = OffsetImage(plt.imread(io.BytesIO(png_data)), zoom=0.9)
            ab = AnnotationBbox(image, (x, y), frameon=False)
            ax.add_artist(ab)
        else:
            print(f"Missing SVG for: {node}")

        drawn_nodes.add(node)

    # Title and axes
    ax.set_title(cuisine.replace("_", " ").title(), y=-0.1, fontsize=50, color="black")
    ax.set_axis_off()
    os.makedirs("figures/type_graphs/max_spanning_trees", exist_ok=True)

    # Save figure (repo-friendly path)
    plt.savefig(f"figures/type_graphs/max_spanning_trees/{cuisine}.svg", bbox_inches="tight")
    plt.show()
```

```python
import pandas as pd
import itertools
import networkx as nx
import numpy as np

# Load data
df = pd.read_csv("ingredient_edgelist.csv")

# Clean columns
df.columns = df.columns.str.strip().str.lower()

# Group by cuisine
cuisines = df["cuisine"].unique()

results = []


import pickle

category_graph_dict = pickle.load(
    open("post_processed_data/nx_category_graph_dict_from_edgelist.pickle", "rb")
)

results = []

for cuisine, G in category_graph_dict.items():

    if G.number_of_edges() == 0:
        continue

    edges = [(u, v, d["weight"]) for u, v, d in G.edges(data=True)]
    edges_sorted = sorted(edges, key=lambda x: x[2], reverse=True)

    total_weight = sum(w for _, _, w in edges_sorted)

    top5 = edges_sorted[:5]
    top50 = edges_sorted[:50]

    top5_fraction = sum(w for _, _, w in top5) / total_weight
    top50_fraction = sum(w for _, _, w in top50) / total_weight

    top5_pairs = [f"{u} - {v}" for u, v, _ in top5]

    results.append([
        cuisine,
        ", ".join(top5_pairs),
        round(top5_fraction, 2),
        round(top50_fraction, 2)
    ])

# Create table
table_df = pd.DataFrame(
    results,
    columns=["Cuisine", "Top 5 Edges", "Top 5 Weight Fraction", "Top 50 Weight Fraction"]
)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_columns', None)
table_df = table_df.sort_values("Cuisine").reset_index(drop=True)
import os
from IPython.display import display, HTML

ICON_FOLDER = "category_symbols"  # folder name

import base64

def normalize(x):
    return x.strip().title()   # VERY IMPORTANT

def pair_to_icons(pair_string):
    parts = pair_string.split(" - ")
    html = ""

    for part in parts:
        filename = normalize(part) + ".svg"
        img_path = os.path.join(ICON_FOLDER, filename)

        if os.path.exists(img_path):
            with open(img_path, "rb") as f:
                encoded = base64.b64encode(f.read()).decode()

            html += f'<img src="data:image/svg+xml;base64,{encoded}" width="28" style="vertical-align:middle;"> '
        else:
            print("❌ Missing:", img_path)   # DEBUG
            html += part + " "

    return html.strip()

def convert_top5_to_icons(edge_string):
    pairs = edge_string.split(",")
    return " , ".join(pair_to_icons(p.strip()) for p in pairs)


# Apply conversion
table_df["Top 5 Edges"] = table_df["Top 5 Edges"].apply(convert_top5_to_icons)

# Convert to HTML table
html_table = table_df.to_html(escape=False, index=False)

display(HTML(html_table))


# display(table_df)   # instead of print(table_df)
```

```python
import pickle
import numpy as np
import pandas as pd
import networkx as nx

# Load category graphs (NOT MST)
category_graph_dict = pickle.load(
    open("post_processed_data/nx_category_graph_dict_from_edgelist.pickle", "rb")
)
def normalize_graph(G):
    total_weight = sum(d["weight"] for _, _, d in G.edges(data=True))
    if total_weight == 0:
        return G
    for u, v, d in G.edges(data=True):
        d["weight"] = d["weight"] / total_weight
    return G
from collections import defaultdict

pair_weights = defaultdict(list)

for cuisine, G in category_graph_dict.items():

    G = normalize_graph(G.copy())

    for u, v, d in G.edges(data=True):

        pair = tuple(sorted([u, v]))  # ensure consistent ordering

        pair_weights[pair].append(d["weight"])
results = []

for pair, weights in pair_weights.items():

    avg_weight = np.mean(weights)
    std_weight = np.std(weights)

    results.append([
        f"{pair[0]} - {pair[1]}",
        avg_weight,
        std_weight
    ])

table_df = pd.DataFrame(
    results,
    columns=["Type Pair", "Average Weight", "St. Dev."]
)
table_df["Average Weight"] = table_df["Average Weight"].round(3)
table_df["St. Dev."] = table_df["St. Dev."].round(3)

top_avg = table_df.sort_values(
    "Average Weight",
    ascending=False
).head(10)
top_std = table_df.sort_values(
    "St. Dev.",
    ascending=False
).head(10)
top_std = top_std[["Type Pair", "St. Dev.", "Average Weight"]]

print("Top 10 by Average Weight")
display(top_avg)

print("Top 10 by Standard Deviation")
display(top_std)
```
